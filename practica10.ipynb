```python
def rec1(n):
    if (n<= 1):
        return 1
    else:
        return rec1 (n-1) + rec1(n-1)
```

**a)** Caso base (if n <= 1): elcodigo hace un simple return 1, esto tarda un tiempo constante T(1) = C1, 
Caso recursivo(Else): el codigo llama a la funcion 2 veces , ambas con el parametro n-1, y luego  suma sus 2 resultados;
Primera llamada: T(n-1), Segunda llamada: T(n-1), La operacion de suma (+) toma un tiempo constante C2:  
*Se cumple La función:*
$$ t(n) = t(n-1) +t(n-1) + C2 $$ = $$ t(n) = 2T(n-1) $$

**b)** 

*Se que $ T(n-1) = 2T((n-1)-1) +c $*
*Tomo : $ t(n) = 2T(n-1) $ y lo agrego donde va $ T(n-1) $*
$$ T(n) = 2[2T(n-2)+c]+c $$
**Aplico propiedad distributiva:multiplico por el 2**
$$ T(n) = 4T(n-2) + 2c + c $$
**Se que $$T(n-2)= 2T(n-3)+c $$**
**Reemplazo T(n-2) = 2T(n-3) en T(n)**
$$ T(n) = 8T(n-3)+7c $$  

**Si miro los coeficientes de los resultados crecen en potencias de bases**

**(1)**$ 2^1 T(n-1)+c(1) $

**(2)**$ 2^2 T(n-2) + c(2+1) $

**(3)**$ 2^3 T(n-3) + c(4+2+1) $

**Si hago esto k-veces (donde k es un numero de pasos cualquiera)**

**La formula universal queda:**

$$ T(n) = 2^k T(n-k)+c(2^(k-1) + ...+ 4+2+1) $$ **Aca es donde aplico la formula:** 

$$ \sum_{i=0}^{n} a^i = 1 + a + a^2 + \ldots + a^n = \frac{a^{n+1}-1}{a-1} $$
**Donde el numero por el que  multiplico a es 2(2  llamadas recursivas) y m es k-1**
**Reemplazo en la formula:**
$$ ((2^(k-1)+1) -1)/(2-1) $$ 
**obtengo 2^k - 1**
**y lo reemplazo en c(2^(k-1) + ...+ 4+2+1)**
$$T(n) = 2^k T(n-k) + c(2^k -1) $$
**Si la recursividad se detiene cuando llego al if (n<=1) mi parametro n-k debe ser igual a 1**
*si n - k = 1 entonces despejando k obtengo k = m - 1*
**Reemplazo todas las k de nuestra ecuacion por n-1**
$$ T(n) = 2^n-1 T(1) + C(2^(n-1)-1) $$

#### Algunas funciones matemáticas útiles


*La función constante:*


$$ f(n) = c $$


Cualquier función constante puede ser escrita como:



$$ \textrm{ si } g(n)=1, \textrm{ entonces } f(n) = c \, g(n). $$



*La función logaritmo:*


$$ f(n) = \log_b(n) $$

La base mas usada es $b= 2$ ya que muchos algoritmos reducen su ejecución reduciendo el problema en dos partes.


*La función lineal:*


$$ f(n) = n $$



*La función N-Log-N:*


$$ f(n) = n \, log(n) $$


*La función cuadratica:*


$$ f(n) = n^2 $$


La función cuadratica generalmente aparece cuando tenemos ciclos <code>for</code> anidados. En la primer iteración realizamos una operación, luego dos, ..., y asi sucecivamente.

$$ 1 + 2 + 3 + \ldots + (n-2) + (n-1) + n \approx \frac{n \, (n+1)}{2} \approx n^2$$


*La función cubica y otros polinomios:*

Cubica:

$$ f(n) = n^3 $$


Polinomios:

$$ f(n) = a_0 + a_1 n + a_2 n^2 + a_3 n^3 + \ldots + a_d n^d $$


Generalemnte el termino de mayor grado es considerado.


Sumas y Series:

$$\sum_{i=a}^{b} f(i) = f(a) + f(a+1) + f(a+2) + \ldots + f(b) $$


$$\sum_{i=1}^{n} i = \frac{n \, (n+1)}{2} $$


Tmabien pdemos escribir los polinomios de esta manera:

$$f(n) = \sum_{i=0}^{d} a_i n^i $$



*La función exponencial:*


$$ f(n) = b^n $$


Sumas Geometricas:

$$\sum_{i=0}^{n} a^i = 1 + a + a^2 + \ldots + a^n = \frac{a^{n+1}-1}{a-1} $$

Ejemplo:

$$ 1 + 2 + 4 + 8 + \ldots + 2^{n-1} = 2^n -1 $$
